1 задача

```python
class TreeNode:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None

def build_tree(arr, i=0):
    if i >= len(arr) or arr[i] is None:
        return None

    root = TreeNode(arr[i])
    root.left = build_tree(arr, 2 * i + 1)
    root.right = build_tree(arr, 2 * i + 2)

    return root

def print_tree(root):
    if root is None:
        return []

    queue = [root]
    result = []

    while queue:
        node = queue.pop(0)
        result.append(node.val)

        if node.left:
            queue.append(node.left)
        if node.right:
            queue.append(node.right)

    return result

arr = [8, 9, 11, 7, 16, 3, 1]
root = build_tree(arr)

print(print_tree(root))
```

Корень дерева находится в начале массива. Для элемента с индексом i левый потомок находится по индексу 2 * i + 1, а правый — по индексу 2 * i + 2. Узлы дерева хранятся в объектах класса TreeNode, у которых есть значение и ссылки на левого и правого потомков.

2 задача

```python
def is_mirror(left, right):
    if left is None and right is None:
        return True

    if left is None or right is None:
        return False

    return (left.val == right.val and
            is_mirror(left.left, right.right) and
            is_mirror(left.right, right.left))

def is_symmetric(root):
    if root is None:
        return True

    return is_mirror(root.left, root.right)

arr = [3, 8, 8, 9, 6, 6, 9]
root = build_tree(arr)

print(is_symmetric(root))
```

Сравниваются узлы, расположенные зеркально: левый потомок левого узла с правым потомком правого узла, а правый потомок левого узла — с левым потомком правого. Значения пар должны совпадать.

3 задача

```python
def min_depth(root):
    if root is None:
        return 0

    if root.left is None and root.right is None:
        return 1

    if root.left is None:
        return 1 + min_depth(root.right)

    if root.right is None:
        return 1 + min_depth(root.left)

    return 1 + min(min_depth(root.left), min_depth(root.right))

arr = [11, 8, 16, 2, 9, None, 6, 7, None, None, 9]
root = build_tree(arr)

print(min_depth(root))
```

BFS проходит дерево по уровням, начиная с корня. Поэтому первый найденный лист будет ближайшим к корню. Значит, глубина этого листа и будет минимальной.

4 задача

```python
def min_max_product(arr):
    if not arr or arr[0] is None:
        return None

    min_index = 0
    while 2 * min_index + 1 < len(arr):
        next_index = 2 * min_index + 1
        if arr[next_index] is None:
            break
        min_index = next_index

    max_index = 0
    while 2 * max_index + 2 < len(arr):
        next_index = 2 * max_index + 2
        if arr[next_index] is None:
            break
        max_index = next_index

    return arr[min_index] * arr[max_index]

arr = [16, 12, 18, 11, 15, 17, 21, 19, 24]

print(min_max_product(arr))
```

В бинарном дереве поиска минимальный элемент находится при движении от корня влево, а максимальный — при движении вправо. Поэтому нужно пройти только по одной ветви дерева. Время поиска составляет O(h), где h — высота дерева.

5 задача

```python
def is_same_tree(a, b):
    if a is None and b is None:
        return True

    if a is None or b is None:
        return False

    if a.val != b.val:
        return False

    return (is_same_tree(a.left, b.left) and
            is_same_tree(a.right, b.right))

arr1 = [16, 12, 18, 15, 17]
arr2 = [16, 12, 18, 15, 17]
arr3 = [16, 11, 18, 15, 17]

tree1 = build_tree(arr1)
tree2 = build_tree(arr2)
tree3 = build_tree(arr3)

print(is_same_tree(tree1, tree2))
print(is_same_tree(tree1, tree3))
```

Чтобы деревья считались одинаковыми, у них должна совпадать структура и значения всех соответствующих узлов. Если на одном и том же месте узел отсутствует только в одном дереве или значения узлов отличаются, деревья не являются одинаковыми.
