Задача 1

```python
n = int(input())

left = 0
right = n

while left <= right:
    mid = (left + right) // 2

    if mid * mid <= n:
        left = mid + 1
    else:
        right = mid - 1

print(right)
```

Используется бинарный поиск. Левая граница равна 0, правая равна n, потому что корень неотрицательного числа не больше самого числа. Если квадрат среднего числа меньше n, ищем правее. Если больше, ищем левее. На каждом шаге диапазон уменьшается примерно в два раза, поэтому время работы составляет O(log n), а дополнительная память - O(1).

Задача 2

```python
n, x, y = map(int, input().split())

left = 0
right = n * min(x, y)

while left < right:
    mid = (left + right) // 2

    if mid // x + mid // y >= n:
        right = mid
    else:
        left = mid + 1

print(left)
```

Сначала самый быстрый ксерокс делает первую копию. Затем оба ксерокса работают оставшееся время.
Проверка достаточности времени выполняется так: складываем количество копий, которые оба ксерокса могут сделать за это время. Если копий достаточно, пробуем уменьшить время. Иначе увеличиваем его.
Бинарный поиск позволяет найти минимальное время. Сложность - O(log(n * max(x, y))).

Задача 3

```python
animals = list(map(int, input().split()))
food = list(map(int, input().split()))

animals.sort()
food.sort()

i = 0
j = 0
count = 0

while i < len(animals) and j < len(food):
    if food[j] >= animals[i]:
        count += 1
        i += 1
    j += 1

print(count)
```

Сначала сортируем животных по возрастанию требуемого количества еды и так же сортируем порции.
Каждый раз пытаемся накормить самое маленькое животное самой маленькой подходящей порцией. Если порция слишком маленькая, переходим к следующей.
Так мы не тратим большую порцию на животное, которому хватило бы маленькой. Сложность - O(n log n + m log m), где n - количество животных, а m - количество порций.

Задача 4

```python
a = input()
b = input()

result = 0

for char in a:
    result ^= ord(char)

for char in b:
    result ^= ord(char)

print(chr(result))
```

Используется операция XOR. Одинаковые символы при применении XOR друг к другу дают ноль. Поэтому все одинаковые буквы из двух строк сокращаются, и остается только добавленная буква.
Сложность - O(n), дополнительная память - O(1).

Задача 5

```python
nums = list(map(int, input().split()))
target = int(input())

seen = {}

for num in nums:
    need = target - num

    if need in seen:
        print([need, num])
        break

    seen[num] = True
else:
    print([])
```

Используется словарь, в котором хранятся уже просмотренные числа.
Для каждого числа вычисляем, какое число нужно добавить, чтобы получить target. Если оно уже есть в словаре, возвращаем пару. Если пары нет, возвращаем пустой список.
Сложность в среднем - O(n) по времени и O(n) по памяти.

Задача 6

```python
strs = input().split()

groups = {}

for word in strs:
    key = ''.join(sorted(word))

    if key not in groups:
        groups[key] = []

    groups[key].append(word)

print(list(groups.values()))
```

Для каждого слова сортируем буквы. Если у двух слов после сортировки получилась одинаковая строка, значит, это анаграммы.
Слова с одинаковым ключом собираются в одну группу.
Сложность - O(n * k log k), где n - количество слов, а k - максимальная длина слова. Дополнительная память - O(n * k).
