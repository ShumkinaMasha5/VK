```python
def two_sum(nums, target):
    left = 0
    right = len(nums) - 1
    
    while left < right:
        first = nums[left]
        second = nums[right]
        current = first + second

        if current == target:
            return [left, right]
        elif current < target:
            left = left + 1
        else:
            right = right - 1

nums = [2, 7, 11, 15]
target = 9
print(two_sum(nums, target))  
```

```python
def reverse_array(arr):
    left = 0
    right = len(arr) - 1

    while left < right:
        temp = arr[left]
        arr[left] = arr[right]
        arr[right] = temp

        left = left + 1
        right = right - 1

arr = [3, 8, 6, 9, 9, 8, 6]
reverse_array(arr)
print(arr)
```

```python
def reverse(arr, l, r):
    left = 0
    right = r - l - 1

    while left < right:
        temp = arr[l + left]
        arr[l + left] = arr[l + right]
        arr[l + right] = temp
        left = left + 1
        right = right - 1


def rotate_array(arr, k):
    n = len(arr)
    k = k % n
    reverse(arr, 0, n)    
    reverse(arr, 0, k)      
    reverse(arr, k, n)      

arr = [1, 2, 3, 4, 5, 6, 7]
k = 3

rotate_array(arr, k)
print(arr)  
```

```python
def merge_sorted(a, b):
    i = 0
    j = 0
    result = []

    while i < len(a) and j < len(b):
        if a[i] < b[j]:
            result.append(a[i])
            i = i + 1
        else:
            result.append(b[j])
            j = j + 1

    while i < len(a):
        result.append(a[i])
        i = i + 1

    while j < len(b):
        result.append(b[j])
        j = j + 1

    return result

a = [1, 3, 5, 7]
b = [2, 4, 6, 8]

print(merge_sorted(a, b))  
```

```python
def merge_sorted(a, b):
    m = len(a) - len(b)  
    i = m - 1             
    j = len(b) - 1        
    k = len(a) - 1       

    while i >= 0 and j >= 0:
        if a[i] > b[j]:
            a[k] = a[i]
            i = i - 1
        else:
            a[k] = b[j]
            j = j - 1
        k = k - 1

    while j >= 0:
        a[k] = b[j]
        j = j - 1
        k = k - 1

a = [1, 3, 5, 7, 0, 0, 0, 0]
b = [2, 4, 6, 8]

merge_sorted(a, b)
print(a)  
```

```python
def min_subarray_len(nums, target):
    left = 0
    current_sum = 0
    min_len = len(nums) + 1  

    for right in range(len(nums)):
        current_sum = current_sum + nums[right]  

        while current_sum >= target:
            current_len = right - left + 1
            if current_len < min_len:
                min_len = current_len
            
            current_sum = current_sum - nums[left]
            left = left + 1

    if min_len > len(nums):
        return 0
    return min_len

nums = [2, 3, 1, 2, 4, 3]
target = 7

print(min_subarray_len(nums, target))  
```

```python
def sort_01(arr):
    left = 0
    right = len(arr) - 1

    while left < right:
        while left < right and arr[left] == 0:
            left = left + 1

        while left < right and arr[right] == 1:
            right = right - 1

        if left < right:
            temp = arr[left]
            arr[left] = arr[right]
            arr[right] = temp

            left = left + 1
            right = right - 1

arr = [1, 0, 1, 0, 1, 0, 0, 1]

sort_01(arr)
print(arr)  
```

```python
def sort_012(arr):
    low = 0
    mid = 0
    high = len(arr) - 1

    while mid <= high:
        if arr[mid] == 0:
            temp = arr[low]
            arr[low] = arr[mid]
            arr[mid] = temp

            low = low + 1
            mid = mid + 1

        elif arr[mid] == 1:
            mid = mid + 1

        else:
            temp = arr[mid]
            arr[mid] = arr[high]
            arr[high] = temp

            high = high - 1

arr = [2, 0, 1, 2, 0, 1, 0, 2, 1]

sort_012(arr)
print(arr)  
```

```python
def move_evens(arr):
    even_pos = 0  

    for i in range(len(arr)):
        if arr[i] % 2 == 0:
            temp = arr[i]       
            j = i
            while j > even_pos:
                arr[j] = arr[j - 1]  
                j = j - 1
            arr[even_pos] = temp    
            even_pos = even_pos + 1

arr = [3, 2, 4, 1, 11, 8, 9]

move_evens(arr)
print(arr)  
```

```python
def move_zeros(arr):
    write_pos = 0

    for i in range(len(arr)):
        if arr[i] != 0:
            temp = arr[i]
            arr[i] = arr[write_pos]
            arr[write_pos] = temp
            write_pos = write_pos + 1

arr = [0, 5, 0, 3, 0, 2, 0]

move_zeros(arr)
print(arr)  
```
