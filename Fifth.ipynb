```python
from collections import deque

class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

def is_complete(root):
    if root is None:
        return True

    queue = deque([root])
    flag = False

    while queue:
        node = queue.popleft()

        if node is None:
            flag = True
        else:
            if flag:
                return False

            queue.append(node.left)
            queue.append(node.right)

    return True


root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)
root.right.left = Node(6)

print(is_complete(root))
```

Используем обход в ширину

```python
import heapq

def merge_arrays(arrays):
    result = []
    heap = []

    for i in range(len(arrays)):
        if len(arrays[i]) > 0:
            heapq.heappush(heap, (arrays[i][0], i, 0))

    while heap:
        value, array_num, index = heapq.heappop(heap)
        result.append(value)

        if index + 1 < len(arrays[array_num]):
            next_value = arrays[array_num][index + 1]
            heapq.heappush(heap, (next_value, array_num, index + 1))

    return result

arrays = [
    [1, 4, 7],
    [2, 5, 8],
    [3, 6, 9]
]

print(merge_arrays(arrays))
```

Сложность алгоритма  O(N log K), где N — общее количество элементов, K — количество массивов.

```python
import heapq

def find_k_smallest(arr, k):
    return heapq.nsmallest(k, arr)[-1]

def find_k_largest(arr, k):
    return heapq.nlargest(k, arr)[-1]


numbers = [7, 2, 9, 4, 1, 6, 8]
k = 3

print("3-й наименьший:", find_k_smallest(numbers, k))
print("3-й наибольший:", find_k_largest(numbers, k))
```

Сложность  O(N log K), где N — общее количество элементов массива, K — номер искомого элемента.

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None


def find_k_smallest(root, k):
    result = []

    def inorder(node):
        if node is None:
            return

        inorder(node.left)
        result.append(node.value)
        inorder(node.right)

    inorder(root)
    return result[k - 1]


def find_k_largest(root, k):
    result = []

    def reverse_inorder(node):
        if node is None:
            return

        reverse_inorder(node.right)
        result.append(node.value)
        reverse_inorder(node.left)

    reverse_inorder(root)
    return result[k - 1]

root = Node(5)
root.left = Node(3)
root.right = Node(8)
root.left.left = Node(2)
root.left.right = Node(4)
root.right.left = Node(7)
root.right.right = Node(9)

k = 3

print("3-й наименьший:", find_k_smallest(root, k))
print("3-й наибольший:", find_k_largest(root, k))
```

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None
        self.balance_factor = 0


def calculate_balance(root):
    if root is None:
        return 0

    left_height = calculate_balance(root.left)
    right_height = calculate_balance(root.right)

    root.balance_factor = left_height - right_height

    return max(left_height, right_height) + 1

root = Node(10)
root.left = Node(5)
root.right = Node(15)
root.left.left = Node(3)
root.left.right = Node(7)

calculate_balance(root)

print("Корень:", root.balance_factor)
print("Левый узел:", root.left.balance_factor)
print("Правый узел:", root.right.balance_factor)
```

balance_factor находится рекурсивно

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None


def mirror(root):
    if root is None:
        return

    root.left, root.right = root.right, root.left

    mirror(root.left)
    mirror(root.right)

root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)

mirror(root)

def print_tree(root):
    if root is None:
        return

    print(root.value, end=" ")
    print_tree(root.left)
    print_tree(root.right)

print_tree(root)
```

В данной задаче снова используем рекурсию, где каждый узел обрабатывается отдельно и потом его поддеревья меняются местами

Сложность — O(N), где N — количество узлов дерева.
